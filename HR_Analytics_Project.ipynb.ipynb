    📊 Attrition by Education Field
</h1>

```python
education_summary = df.groupby("EducationField").agg(
    Total_Employees=("Attrition", "count"),
    Employees_Left=("Attrition", lambda x: (x == "Yes").sum())
)

education_summary["Attrition_Rate"] = (
    education_summary["Employees_Left"]
    / education_summary["Total_Employees"] * 100
).round(2)

display(
    education_summary.sort_values(
        "Attrition_Rate", ascending=False
    )
)
```

```python
import matplotlib.pyplot as plt

education_sorted = education_summary["Attrition_Rate"].sort_values(
    ascending=False
)

plt.figure(figsize=(10, 6))

ax = education_sorted.plot(
    kind="bar",
    color=["#E74C3C", "#E67E22", "#3498DB", "#2ECC71", "#9B59B6"],
    edgecolor="black"
)

plt.title(
    "HR Insight: Attrition Rate by Education Field",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("Education Field")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=20, ha="right")

for bar in ax.patches:
    height = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.3,
        f"{height:.2f}%",
        ha="center",
        fontsize=9,
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```
